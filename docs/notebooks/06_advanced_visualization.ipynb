# Tutorial 6: Advanced Visualization

This tutorial covers advanced visualization techniques for exploring regulatory regions in genomic context and examining sequence-level TF binding patterns.

```{note}
Sequence visualization (logos, ISM) requires optional dependencies:
`pip install deepscenic[sequence]`
```

## Setup

````python
import torch

import deepscenic as ds

print(f"deepSCENIC version: {ds.__version__}")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
````

## Load Model and Data

````python
data_dir = "data/MM_lines/"

# Load trained model and data
model = ds.tl.load_model(data_dir + "model.pt", device=device)
mdata = ds.read(data_dir + "preprocessed_data.h5mu")

print(f"Model loaded")
````

## Genome Browser

Multi-track genome browser showing accessibility, genes, and regulatory links:

````python
# Genome browser view around a gene of interest
ds.pl.genome_browser(
    mdata,
    gene="MITF",  # Or specify coordinates: chrom="chr3", start=69700000, end=70200000
)
````

## Arc Plots

Visualize region-gene connections as arcs (Signac-style):

````python
# Arc plot showing region-gene regulatory connections
# Links are colored by E2 weight strength
ds.pl.arc_plot(
    model,
    mdata,
    gene="MITF",
    top_k=20,  # Show top 20 regulatory links
)
````

## Sequence Visualization

Visualize TF binding patterns from sequence attributions.

### Sequence Logos

Sequence logos show which nucleotides matter for TF binding:

````python
# Attribution-based logo showing which positions matter for TF binding
ds.pl.logo_attribution(
    model,
    region="chr3:69800000-69801000",
    tf_name="SOX10",
)
````

````python
# Motif logo from learned weights
ds.pl.logo_motif(model, tf_name="SOX10")
````

### In-Silico Mutagenesis

ISM heatmaps show which nucleotides are important for TF binding:

````python
# In-silico mutagenesis heatmap
# Shows effect of each possible mutation on TF binding
ds.pl.ism_heatmap(
    model,
    region="chr3:69800000-69800100",  # Short region for clarity
    tf_name="SOX10",
)
````

## Exporting Figures

Save publication-quality figures:

````python
# Save figure directly
ds.pl.heatmap_grn(model, save="grn_heatmap.pdf")

# Or get figure object for more control
fig = ds.pl.heatmap_grn(model, show=False, return_fig=True)
fig.savefig("grn_heatmap.pdf", dpi=300, bbox_inches="tight")
````
