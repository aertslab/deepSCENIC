# Predict embeddings

```python
DATA_PATH = '/lustre1/project/stg_00002/lcb/gpartel/projects/deepSCENIC/data/DPCL/'

ad_atac = sc.read(DATA_PATH + 'fragment_matrix.h5ad')
regions = ad_atac.var_names
```

```python
from src.deepSCENIC import build_seq_dataloader

ad_atac.X[ad_atac.X>0]=1
dataloader, _, _, _ = build_seq_dataloader(opt, ad=ad_atac, pretrain=True)
```

```python
import h5py, os

h5_name = DATA_PATH + '/Sei_embs.h5'
print(not os.path.exists(h5_name))
if not os.path.exists(h5_name):
    f = h5py.File(h5_name, "w")

    ds_X = f.create_dataset(
        "X",
        (len(regions), opt.emb_len, opt.TF2rNet_bottleneck_size),
        dtype="float32",
    )
```

```python
from src.tf2rNet.models import TF2rNet, MotifNet, Sei

tf2rNet_func_encoder = Sei().float().to(opt.device)
model_dict_tf2r_func_enc = torch.load(opt.save_name + 'TF2rNET_func_enc_pretrain.pth',  map_location=torch.device(opt.device))['model_state_dict']
tf2rNet_func_encoder.load_state_dict(model_dict_tf2r_func_enc)
```

```python
from tqdm import tqdm
from torch.autograd import Variable
Tensor = torch.cuda.FloatTensor
    
tf2rNet_func_encoder.eval()
embs = []
i = 0
with torch.no_grad():
    for i, (data_batch, data_batch_idx) in tqdm(enumerate(dataloader, 0), unit="batch", total=len(dataloader)):
        X,y = data_batch
        X = X.to(opt.device)
        emb = tf2rNet_func_encoder(X)  
        ds_X[i:i+emb.shape[0]] = emb.cpu().numpy().astype('float32')
        i = i+emb.shape[0]

f.close()
```
